```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,DecisionTreeRegressor
from sklearn.metrics import accuracy_score, confusion_matrix, mean_absolute_error,r2_score,classification_report
```

```python
df = pd.read_csv('Metro_Interstate_Traffic_Volume.csv')
```

```python
df
```

```python
df = df.sample(10000,random_state=42)
```

```python
df['holiday'] = df['holiday'].fillna('Normal day')
```

```python
df['holiday'].value_counts()
```

```python
df['holiday'].isnull().sum()
```

```python
x = df.drop(columns = ['traffic_volume'])
y = df['traffic_volume']
```

```python
x = pd.get_dummies(x)
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
dcr = DecisionTreeRegressor(max_depth=10,random_state=42)
```

```python
dcr.fit(xtrain, ytrain)
```

```python
ypred = dcr.predict(xtest)
```

```python
print("R2 Score:", r2_score(ytest, ypred))
print("MAE:", mean_absolute_error(ytest, ypred))
```
